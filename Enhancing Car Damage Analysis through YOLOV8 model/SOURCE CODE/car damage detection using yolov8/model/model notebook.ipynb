```python
import torch
import ultralytics
from ultralytics import YOLO
import os
```

```python
print("Torch version:", torch.__version__)
print("CUDA available:", torch.cuda.is_available())
```

```python
dataset_path = "/kaggle/input/dataset"
print(os.listdir(dataset_path))
```

```python
# Load YOLO model (change 'yolov8s.pt' to 'yolov8m.pt' or 'yolov8l.pt' if needed)
model = YOLO("yolov8s.pt")

# Train the model
model.train(
    data=f"{dataset_path}/data.yaml",  # Path to data.yaml
    epochs=200,  # Adjust epochs as needed
    imgsz=640,  # Image size (default: 640)
    batch=16,  # Adjust batch325 size based on GPU memory
    workers=2,  # Number of workers for data loading
    project="/kaggle/working/",  # Output directory
    name="yolov8_train"  # Name of the run
)
```
