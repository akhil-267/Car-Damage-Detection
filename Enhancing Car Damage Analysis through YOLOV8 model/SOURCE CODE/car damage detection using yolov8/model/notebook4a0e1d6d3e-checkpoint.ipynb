```python
import torch
import ultralytics
from ultralytics import YOLO
import os
```

```python
print("Torch version:", torch.__version__)
print("CUDA available:", torch.cuda.is_available())
```

```python
import os

dataset_path = "/kaggle/input/d/akhilmuthyam/dataset"
print(os.listdir(dataset_path))
```

```python
# Load YOLO model (change 'yolov8s.pt' to 'yolov8m.pt' or 'yolov8l.pt' if needed)
model = YOLO("yolov8s.pt")

# Train the model
model.train(
    data=f"{dataset_path}/data.yaml",  # Path to data.yaml
    epochs=100,  # Adjust epochs as needed
    imgsz=640,  # Image size (default: 640)
    batch=16,  # Adjust batch size based on GPU memory
    workers=2,  # Number of workers for data loading
    project="/kaggle/working/",  # Output directory
    name="yolov8_train"  # Name of the run
)
```

```python
import os

trained_model_path = "/kaggle/working/yolov8_train/weights/"

if os.path.exists(trained_model_path):
    print("Weights folder contents:", os.listdir(trained_model_path))
else:
    print("❌ Weights folder not found!")
```

```python
for root, dirs, files in os.walk("/kaggle/working/"):
    for file in files:
        if file.endswith(".pt"):
            print(f"Found model file: {os.path.join(root, file)}")
```
